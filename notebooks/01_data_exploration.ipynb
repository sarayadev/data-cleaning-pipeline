• The dataset contains more female than male transactions, with females accounting for approximately 60% of transactions and males accounting for approximately 40%.

• Cash is the most frequently used payment method, while debit cards are the least frequently used.

• Mall of Istanbul, Kanyon, and Metrocity have the highest number of transactions among the shopping malls in the dataset.

• Customer ages range from 18 to 69, with an average age of approximately 43 years.

• The average purchase price is $689.26, while the median is $203.30. The difference between these values suggests that higher-priced purchases may be influencing the average.

• Customer IDs and invoice numbers appear to function primarily as identifiers rather than variables for numerical analysis.

• Invoice dates may be useful for analyzing purchasing trends over time.
