# Data Cleaning Findings

After inspecting the dataset for missing values, duplicate records, data types, and inconsistent categorical values, I found the following:

- **Missing values:** None detected.
- **Duplicate records:** None detected.
- **Inconsistent values:** None detected in the categorical columns.
- **Data types:** The `invoice_date` column is stored as an `object` instead of a `datetime` data type.

## Conclusion

The dataset appears to be clean in terms of missing values, duplicates, and inconsistent categorical values. The primary issue identified is the `invoice_date` data type, which should be converted to `datetime` during the cleaning process.


```python
df['invoice_date'] = pd.to_datetime(df['invoice_date'], format='%d/%m/%Y')
```
